# Hybrid BM25 + RAG
Запуск гибридного поиска аналогов и генерация LLM-отчетов прямо из ноутбука.

```python
import sys
import logging
from pathlib import Path

# 1. Настройка путей и импортов
# Добавляем родительскую директорию в путь, чтобы видеть основной код
notebook_path = Path().absolute()
root_dir = notebook_path.parent
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))

from hybrid_bm25_rag import HybridConfig, HybridPipeline
from rag_core.constants import PROJECT_ROOT

# 2. Настройка логов для вывода в ноутбук
logging.basicConfig(
    level=logging.INFO, 
    format="%(message)s",
    stream=sys.stdout,
    force=True
)

print(f"Корень проекта: {PROJECT_ROOT}")
```

```python
# 3. Конфигурация запуска
config = HybridConfig(
    dataset=PROJECT_ROOT / "Best-Buy-dataset-clean.csv",
    indices=(0, 10, 20),       # Список индексов товаров
    top_k=5,                   # Сколько аналогов искать
    gen_report=True,           # Генерировать ли отчет через LLM
    llm_model="qwen",          # Модель qwen (0.5B)
    llm_backend="transformers",
    hf_allow_download=False,   # Использовать только локальный кеш
    report_dir=PROJECT_ROOT / "reports"
)

# 4. Запуск пайплайна
pipeline = HybridPipeline(config)
pipeline.run()
```
